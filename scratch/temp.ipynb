```python
df = pd.read_csv('/Users/stjia/Documents/cloze-encounters/filedata/book-level.csv')
```

```python
score_cols = ['gpt35_score', 'gpt4o_score', 'claude_score', 'llama8b_score', 'llama70b_score', 'gemini_score']
avg_scores_by_books3 = df.groupby('in_books3')[score_cols].mean()
avg_scores_by_books3
```

```python
# Calculate the average score across all models for each group in avg_scores_by_books3
avg_scores_by_books3['avg_model_score'] = avg_scores_by_books3[score_cols].mean(axis=1)
avg_scores_by_books3[['avg_model_score']]
```
